```python
from pyspark.sql import functions as F

df_clean = df.select(
    F.col("Order ID").alias("order_id"),
    F.col("Date").alias("order_date"),
    F.col("Status").alias("order_status"),
    F.col("Fulfilment").alias("fulfilment"),
    F.col("Sales Channel ").alias("sales_channel"),
    F.col("ship-service-level").alias("ship_service_level"),
    F.col("Style").alias("style"),
    F.col("SKU").alias("sku"),
    F.col("Category").alias("category"),
    F.col("Size").alias("size"),
    F.col("ASIN").alias("asin"),
    F.col("Courier Status").alias("courier_status"),
    F.col("Qty").alias("qty"),
    F.col("currency").alias("currency"),
    F.col("Amount").alias("amount"),
    F.col("ship-city").alias("ship_city"),
    F.col("ship-state").alias("ship_state"),
    F.col("ship-postal-code").alias("ship_postal_code"),
    F.col("ship-country").alias("ship_country"),
    F.col("promotion-ids").alias("promotion_ids"),
    F.col("B2B").alias("b2b"),
    F.col("fulfilled-by").alias("fulfilled_by")
)

```

#converting datatypes 

```python
df_clean = (
    df_clean
    .withColumn(
        "order_date",
        F.to_date("order_date", "MM-dd-yy")
    )
    .withColumn("qty", F.col("qty").cast("integer"))
    .withColumn("amount", F.col("amount").cast("double"))
)
```

%md handling missing values

```python
df_clean = df_clean.fillna({
    "order_status":"Unknown",
    "courier_status":"Unknown",
    "category": "Unknown",
    "currency": "Unknown",
    "ship_country": "Unknown"
})
```

```python
df_clean = df_clean.fillna({
    "qty":0,
    "amount":0.0
})
```

fixing state names and region follow up

```python
df_clean = df_clean.withColumn(
    "ship_state",
    F.when(
        F.upper(F.trim(F.col("ship_state"))).isin(
            "RAJASTHAN",
            "RAJSHTHAN",
            "RAJSTHAN",
            "RJ"
        ),
        "RAJASTHAN"
    ).when(
        F.upper(F.trim(F.col("ship_state"))).isin(
            "ODISHA",
            "ORISSA"
        ),
        "ODISHA"
    )
    .when(
        F.upper(F.trim(F.col("ship_state"))).isin(
            "DELHI",
            "NEW DELHI"
        ),
        "DELHI"
    )
    .when(
        F.upper(F.trim(F.col("ship_state"))).isin(
            "PUDUCHERRY",
            "PONDICHERRY"
        ),
        "PUDUCHERRY"
    )
    .when(
        F.upper(F.trim(F.col("ship_state"))).isin(
            "PB",
            "PUNJAB/MOHALI/ZIRAKPUR"
        ),
        "PUNJAB"
    )
    .when(
        F.upper(F.trim(F.col("ship_state"))) == "NL",
        "NAGALAND"
    )
    .when(
        F.upper(F.trim(F.col("ship_state"))) == "AR",
        "ARUNACHAL PRADESH"
    )
    .otherwise(
        F.upper(F.trim(F.col("ship_state")))
    )
)
```

```python
display(
    df_clean
    .select("ship_state")
    .distinct()
    .orderBy("ship_state")
)
```

```python
df_clean = df_clean.withColumn(
    "region",
    F.when(
        F.col("ship_state").isin(
            "JAMMU & KASHMIR",
            "HIMACHAL PRADESH",
            "PUNJAB",
            "CHANDIGARH",
            "UTTARAKHAND",
            "HARYANA",
            "DELHI",
            "RAJASTHAN",
            "UTTAR PRADESH",
            "LADAKH"
        ),
        "North"
    )
    .when(
        F.col("ship_state").isin(
            "MAHARASHTRA",
            "GUJARAT",
            "GOA",
            "MADHYA PRADESH",
            "CHHATTISGARH",
            "DADRA AND NAGAR"
        ),
        "West"
    )
    .when(
        F.col("ship_state").isin(
            "WEST BENGAL",
            "ODISHA",
            "BIHAR",
            "JHARKHAND",
            "SIKKIM"
        ),
        "East"
    )
    .when(
        F.col("ship_state").isin(
            "ASSAM",
            "ARUNACHAL PRADESH",
            "MANIPUR",
            "MEGHALAYA",
            "MIZORAM",
            "NAGALAND",
            "TRIPURA"
        ),
        "Northeast"
    )
    .when(
        F.col("ship_state").isin(
            "ANDHRA PRADESH",
            "KARNATAKA",
            "KERALA",
            "TAMIL NADU",
            "TELANGANA",
            "PUDUCHERRY",
            "ANDAMAN & NICOBAR",
            "LAKSHADWEEP"
        ),
        "South"
    )
    .otherwise("Other")
)
```

```python
from pyspark.sql import functions as F

# ============================================================
# 1. Read the two additional files
# ============================================================

international_df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv("Files/International sale Report.csv")
)

sale_df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv("Files/Sale Report.csv")
)

# ============================================================
# 2. Display their schemas
# ============================================================

print("========== INTERNATIONAL SALE REPORT ==========")
international_df.printSchema()

print("\n========== SALE REPORT ==========")
sale_df.printSchema()

# ============================================================
# 3. Create standardized SKU lists
# ============================================================

amazon_skus = (
    df_clean
    .select(
        F.upper(F.trim(F.col("sku"))).alias("sku")
    )
    .filter(
        F.col("sku").isNotNull() &
        (F.col("sku") != "")
    )
    .distinct()
)

international_skus = (
    international_df
    .select(
        F.upper(F.trim(F.col("SKU"))).alias("sku")
    )
    .filter(
        F.col("sku").isNotNull() &
        (F.col("sku") != "")
    )
    .distinct()
)

sale_skus = (
    sale_df
    .select(
        F.upper(F.trim(F.col("SKU Code"))).alias("sku")
    )
    .filter(
        F.col("sku").isNotNull() &
        (F.col("sku") != "")
    )
    .distinct()
)

# ============================================================
# 4. Count unique SKUs
# ============================================================

amazon_count = amazon_skus.count()
international_count = international_skus.count()
sale_count = sale_skus.count()

# ============================================================
# 5. Find SKU overlap
# ============================================================

amazon_international = (
    amazon_skus
    .join(international_skus, "sku", "inner")
)

amazon_sale = (
    amazon_skus
    .join(sale_skus, "sku", "inner")
)

international_sale = (
    international_skus
    .join(sale_skus, "sku", "inner")
)

# ============================================================
# 6. Display results
# ============================================================

print("\n========== SKU COUNTS ==========")
print("Amazon Sale Report SKUs      :", amazon_count)
print("International Sale Report SKUs:", international_count)
print("Sale Report SKUs             :", sale_count)

print("\n========== SKU OVERLAP ==========")
print(
    "Amazon ↔ International       :",
    amazon_international.count()
)

print(
    "Amazon ↔ Sale Report         :",
    amazon_sale.count()
)

print(
    "International ↔ Sale Report  :",
    international_sale.count()
)

# ============================================================
# 7. Show some matching SKUs
# ============================================================

print("\n========== AMAZON + INTERNATIONAL MATCHES ==========")
display(amazon_international.limit(20))

print("\n========== AMAZON + SALE REPORT MATCHES ==========")
display(amazon_sale.limit(20))
```

```python
from pyspark.sql import functions as F

# Read May-2022 product file
may_df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv("Files/May-2022.csv")
)

# Show schema
may_df.printSchema()

# Standardize Amazon SKUs
amazon_skus = (
    df_clean
    .select(
        F.upper(F.trim(F.col("sku"))).alias("sku")
    )
    .filter(
        F.col("sku").isNotNull() &
        (F.col("sku") != "")
    )
    .distinct()
)

# Standardize May-2022 SKUs
may_skus = (
    may_df
    .select(
        F.upper(F.trim(F.col("Sku"))).alias("sku")
    )
    .filter(
        F.col("sku").isNotNull() &
        (F.col("sku") != "")
    )
    .distinct()
)

# Find matching SKUs
matching_skus = amazon_skus.join(
    may_skus,
    "sku",
    "inner"
)

print("Amazon unique SKUs :", amazon_skus.count())
print("May-2022 unique SKUs:", may_skus.count())
print("Matching SKUs      :", matching_skus.count())

display(matching_skus.limit(20))
```

```python
from pyspark.sql import functions as F

# Rename problematic columns first
sale_clean = (
    sale_df
    .withColumnRenamed("SKU Code", "sku_code")
    .withColumnRenamed("Design No.", "design_no")
    .withColumnRenamed("Category", "category")
    .withColumnRenamed("Size", "size")
    .withColumnRenamed("Color", "color")
    .withColumnRenamed("Stock", "stock")
)

# Check the new schema
sale_clean.printSchema()

# Find duplicate SKU records
sale_duplicates = (
    sale_clean
    .groupBy("sku_code")
    .agg(
        F.count("*").alias("records"),
        F.countDistinct("design_no").alias("designs"),
        F.countDistinct("category").alias("categories"),
        F.countDistinct("size").alias("sizes"),
        F.countDistinct("color").alias("colors")
    )
    .filter(F.col("records") > 1)
)

print(
    "SKU codes with multiple records:",
    sale_duplicates.count()
)

display(
    sale_duplicates
    .orderBy(F.desc("records"))
    .limit(20)
)
```

```python
# Check how many unique SKU codes exist
total_skus = sale_clean.select("sku_code").distinct().count()

# Check how many SKU codes occur more than once
duplicate_skus = (
    sale_clean
    .groupBy("sku_code")
    .count()
    .filter(F.col("count") > 1)
)

print("Total unique SKU codes:", total_skus)
print("SKU codes with duplicates:", duplicate_skus.count())

display(
    duplicate_skus
    .orderBy(F.desc("count"))
    .limit(20)
)
```

```python
amazon_skus = (
    df_clean
    .select(F.upper(F.trim(F.col("sku"))).alias("sku"))
    .filter(
        F.col("sku").isNotNull() &
        (F.col("sku") != "")
    )
    .distinct()
)

sale_skus = (
    sale_clean
    .select(F.upper(F.trim(F.col("sku_code"))).alias("sku"))
    .filter(
        F.col("sku").isNotNull() &
        (F.col("sku") != "")
    )
    .distinct()
)

matching_skus = amazon_skus.join(
    sale_skus,
    "sku",
    "inner"
)

print("Amazon SKUs:", amazon_skus.count())
print("Sale Report SKUs:", sale_skus.count())
print("Matching SKUs:", matching_skus.count())
```

```python
product_lookup = (
    sale_clean
    .select(
        F.upper(F.trim(F.col("sku_code"))).alias("sku"),
        F.col("design_no"),
        F.col("category").alias("product_category"),
        F.col("size").alias("product_size"),
        F.col("color"),
        F.col("stock")
    )
    .filter(
        F.col("sku").isNotNull() &
        (F.col("sku") != "")
    )
    .dropDuplicates(["sku"])
)

print("Product lookup rows:", product_lookup.count())

display(product_lookup.limit(20))
```

```python
product_lookup.groupBy("sku").count().filter(F.col("count") > 1).show()
```

enrichment Join Amazon sales with the product lookup


```python
enriched_df = (
    df_clean
    .join(
        product_lookup,
        F.upper(F.trim(df_clean["sku"])) == product_lookup["sku"],
        "left"
    )
    .select(
        df_clean["*"],
        product_lookup["design_no"],
        product_lookup["product_category"],
        product_lookup["product_size"],
        product_lookup["color"],
        product_lookup["stock"]
    )
)

display(enriched_df.limit(10))
```

```python
total_rows = enriched_df.count()

matched_rows = (
    enriched_df
    .filter(F.col("color").isNotNull())
    .count()
)

print("Total sales rows:", total_rows)
print("Rows enriched with product data:", matched_rows)
```

```python
display(
    enriched_df
    .filter(F.col("color").isNull())
    .select(
        "sku",
        "style",
        "category",
        "size"
    )
    .distinct()
    .limit(20)
)
```

Calculate Monthly Revenue

```python
monthly_revenue = (
    enriched_df
    .groupBy(
        F.year("order_date").alias("year"),
        F.month("order_date").alias("month")
    )
    .agg(
        F.sum("amount").alias("monthly_revenue")
    )
    .orderBy("year", "month")
)

display(monthly_revenue)
```

Region-wise Sales

```python
regional_sales = (
    enriched_df
    .groupBy("region")
    .agg(
        F.sum("amount").alias("total_sales")
    )
    .orderBy(F.desc("total_sales"))
)

display(regional_sales)
```

Total Orders and Units

```python
kpi_summary = enriched_df.select(
    F.countDistinct("order_id").alias("total_orders"),
    F.sum("qty").alias("total_units_sold"),
    F.sum("amount").alias("total_revenue")
)

display(kpi_summary)
```

Average order value

```python
aov = enriched_df.select(
    (
        F.sum("amount") /
        F.countDistinct("order_id")
    ).alias("average_order_value")
)

display(aov)
```

Additional KPI: Average Selling Price

```python
average_selling_price = enriched_df.select(
    (
        F.sum("amount") /
        F.sum("qty")
    ).alias("average_selling_price")
)

display(average_selling_price)
```

additional KPI Cancellation Rate

```python

total_orders = (
    enriched_df
    .select(F.countDistinct("order_id").alias("total_orders"))
    .first()["total_orders"]
)

cancelled_orders = (
    enriched_df
    .filter(
        F.upper(F.trim(F.col("order_status"))).contains("CANCEL")
    )
    .select(F.countDistinct("order_id").alias("cancelled_orders"))
    .first()["cancelled_orders"]
)

cancellation_rate = (
    cancelled_orders / total_orders * 100
)

print("Total Orders      :", total_orders)
print("Cancelled Orders  :", cancelled_orders)
print("Cancellation Rate :", round(cancellation_rate, 2), "%")
```

KPI summary 

```python
kpi_summary = enriched_df.select(
    F.countDistinct("order_id").alias("total_orders"),
    F.sum("qty").alias("total_units_sold"),
    F.sum("amount").alias("total_revenue")
).withColumn(
    "average_order_value",
    F.col("total_revenue") / F.col("total_orders")
).withColumn(
    "average_selling_price",
    F.col("total_revenue") / F.col("total_units_sold")
).withColumn(
    "cancellation_rate",
    F.lit(round(cancellation_rate, 2))
)

display(kpi_summary)
```

```python
#Create product-level sales
product_sales = (
    enriched_df
    .groupBy(
        "sku",
        "category"
    )
    .agg(
        F.sum("qty").alias("total_units_sold"),
        F.sum("amount").alias("total_revenue")
    )
    .orderBy(F.desc("total_revenue"))
)

display(product_sales.limit(20))
```
