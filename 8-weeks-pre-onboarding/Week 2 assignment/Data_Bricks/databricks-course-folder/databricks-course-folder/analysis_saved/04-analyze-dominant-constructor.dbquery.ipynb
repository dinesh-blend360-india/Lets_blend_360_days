```sql

with constructor_metrices as (
select constructor_name, 
sum(race_starts) as race_starts,    
sum(number_of_wins) as number_of_wins, 
sum(number_of_podiums) as number_of_podium,
sum(case when standing=1 then 1 else 0 end) total_champions
from 
formulae1.gold.onstructors_standing

group by constructor_name 
having total_champions>=1

)
select constructor_name, race_starts, number_of_wins, number_of_podium, total_champions, (total_champions * 100) + (number_of_wins *10) + (number_of_podium * 3)  as greatness_score from constructor_metrices order by greatness_score;
```
